## Reshaping, stacking, squeezing and unsqueezing 

#### Reshape : reshapes an input tensor to a defined shape
#### View : return a view of an input tensor of certain shape but keep the same memory as the original tensor
#### Stacking : combine multiple tensors on top of each other or side by side same as (**torch.cat**)
#### Squeeze : removes all `1` dimensions from a tensor 
#### Unsqueeze : add a `1` dimension to a target tensor 
#### Permute : Return a view of the input with dimensions permuted in a certain way

```python
y = torch.arange(1., 10.)
y
```

```python
y.shape
```

```python
# Add an extra dimension 
y_reshaped = y.reshape(1, 9) # reshape has to be of original size
y_reshaped, y_reshaped.shape
```

```python
# Chagne the view
z = y.view(1, 9) # z shares the same memory so changing z changes y because it shares the same memory 
z, z.shape
```

```python
z[:, 0] = 5 # see changing z also changes original y
z, y
```

```python
# Stack tensors on top 
y_stacked = torch.stack([y,y,y,y])
y_stacked
```

```python
y_stacked = torch.cat([y,y])
y_stacked
```

```python
# Squeeze tensor
y_reshaped
```

```python
y_reshaped.shape
```

```python
y_reshaped.squeeze()
```

```python
y_reshaped.squeeze().shape
```

```python
# Permute - shifts axis
x_original = torch.rand(size=(224,224,3))   # height, width and color channels 

x_permuted = x_original.permute(2,0,1)    # index 2 = color channel, index 0 = height, index 1 = width 
print(x_original.shape)
print(x_permuted.shape)
```


# Reproducibility - reduce randomness 

```python
torch.rand(3,3)
```

```python
rand_tensorA = torch.rand(3,4)
rand_tensorB = torch.rand(3,4)

print(rand_tensorA)
print(rand_tensorB)
print(rand_tensorA == rand_tensorB)
```

```python
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

rand_tensorC = torch.rand(4,5)
torch.manual_seed(RANDOM_SEED)

rand_tensorD = torch.rand(4,5)
torch.manual_seed(RANDOM_SEED)

print(rand_tensorC)
print(rand_tensorD)
print(rand_tensorC == rand_tensorD)
```
